```python
import numpy as np 
import cv2 as cv
import matplotlib.pyplot as plt
```

```python
image = cv.imread('E:\\New folder (6)\\F-35A_flight.jpg',0)
template = cv.imread('E:\\New folder (6)\\F-35A_flight1.jpg',0)
```

```python
w, h = template.shape
result = cv.matchTemplate(image, template, cv.TM_CCOEFF_NORMED)
```

```python
threshold = 0.6
locations = np.where(result >=threshold)

for point in zip(*locations[::-1]):
    cv.rectangle(image, point, (point[0]+h, point[1]+w), (255,255,0), 2)
```

```python
plt.imshow(image)
plt.show()
```

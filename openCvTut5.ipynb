```python
import numpy as np
import cv2 as cv
import matplotlib.pyplot as plt
```

```python
image = cv.imread('E:\\New folder (6)\\Self-driving-in-Norway.jpg')
```

```python
#SIFT
feat_sift = cv.xfeatures2d.SIFT_create()
```

```python
#SURF
# feat_surf = cv.xfeatures2d.SURF_create()
```

```python
#ORB
feat_orb = cv.ORB_create(nfeatures=1000)
```

```python
sift_keypoints, descriptors = feat_sift.detectAndCompute(image, None)
# sruf_keypoints, descriptors = feat_surf.detectAndCompute(image, None)
orb_keypoints, descriptors = feat_orb.detectAndCompute(image, None)

image_sift = cv.drawKeypoints(image, sift_keypoints, None)
```

```python
image_sift = cv.drawKeypoints(image, sift_keypoints, None)
# image_surf = cv.drawKeypoints(image, surf_keypoints, None)
image_orb = cv.drawKeypoints(image, orb_keypoints, None)
```

```python
plt.imshow(image_sift)
plt.show()
```

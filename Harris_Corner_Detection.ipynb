```python
import numpy as np 
import cv2 as cv
import matplotlib.pyplot as plt
```

```python
image = cv.imread("E:\\New folder (6)\\geometric-shapes-in-art.jpg")
image_gr = cv.imread("E:\\New folder (6)\\geometric-shapes-in-art.jpg",0)
```

```python
corners = cv.cornerHarris(image_gr, 3, 7, 0.04)
corner_dilated = cv.dilate(corners, None)
```

```python
image[corner_dilated > 0.01*corner_dilated.max()] = [255,0,0]
```

```python
plt.imshow(image)
plt.show()
```

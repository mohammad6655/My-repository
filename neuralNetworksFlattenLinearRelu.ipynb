```python
import numpy as np 
import torch
from torch import nn
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

```python
input_image = torch.rand(3, 28, 28) # B: 3
flat_1 = nn.Flatten()

print(callable(flat_1))

flat_image = flat_1(input_image)

print(input_image.shape)
print(flat_image.shape)
```

```python
linear_1 = nn.Linear(in_features = 784, out_features = 100)

print(callable(linear_1))

linear_1_images = linear_1(flat_image)

print(linear_1_images.shape)
```

```python
relu_1 = nn.ReLU()

relu_images = relu_1(linear_1_images)

print(relu_images.shape)
```

```python
input_image = torch.rand(3, 28, 28)

seq_layers = nn.Sequential(
    nn.Flatten(),
    nn.Linear(784, 20),
    nn.ReLU(),
    nn.Linear(20, 10)
)

logits = seq_layers(input_image)

print(logits.shape)
```

```python
softmax = nn.Softmax(1)

preds = softmax(logits)

print(logits)
print('----------------------------------------------------------------------------------')
print(preds)
```

```python
class NeuralNetwork(nn.Module):

    def __init__(self):
        super().__init__()
        self.flatten = nn.Flatten()
        self.linear_relu_stack = nn.Sequential(
            nn.Linear(28*28, 512),
            nn.ReLU(),
            nn.Linear(512, 512),
            nn.ReLU(),
            nn.Linear(512, 10),
        )

    def forward(self, x):
        x = self.flatten(x)
        logits = self.linear_relu_stack(x)
        return logits


my_neural_network = NeuralNetwork().to(device)

print(my_neural_network)
```

```python
data = torch.rand(1, 28, 28, device = device)
logits = my_neural_network(data)
pred_probab = nn.Softmax(dim = 1)(logits)

print(pred_probab)
```

```python
y_pred = pred_probab.argmax(1)
print(f"Predicted class: {y_pred}")
```

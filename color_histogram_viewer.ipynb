```python
import numpy as np 
import cv2 as cv
import matplotlib.pyplot as plt
```

```python
image = cv.imread("E:\\New folder (6)\\F-35A_flight.jpg")
```

```python
w, h, c = image.shape
```

```python
mask = np.zeros(image.shape[0:2], np.uint8)
mask[0:int(w/2), 0:int(h/2)] = 255
```

```python
colors = ['b','g','r']
for i, col in enumerate(colors):
    histogram = cv.calcHist([image], [i], mask, [256], [0, 256])
    plt.plot(histogram, color = col)
```

```python
plt.subplot(121), plt.imshow(image, 'gray')
plt.subplot(122), plt.plot(histogram)
plt.show()
```

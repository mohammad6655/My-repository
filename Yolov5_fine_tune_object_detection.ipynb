import os
import shutil
from random import shuffle

# Source and destination paths
source_path = r"E:\Datasets\Traffic Signs Dataset in YOLO format Sorce\ts"

train_image_path = r"E:\Datasets\Traffic Signs Dataset in YOLO format\train\images"
train_label_path = r"E:\Datasets\Traffic Signs Dataset in YOLO format\train\labels"
val_image_path   = r"E:\Datasets\Traffic Signs Dataset in YOLO format\validation\images"
val_label_path   = r"E:\Datasets\Traffic Signs Dataset in YOLO format\validation\labels"

# Train/validation split ratios
train_ratio = 0.8
val_ratio   = 0.2

# Create directories if they don’t exist
for path in [train_image_path, train_label_path, val_image_path, val_label_path]:
    os.makedirs(path, exist_ok=True)

# Collect all image files
images = [f for f in os.listdir(source_path) if f.endswith((".jpg", ".png"))]

# Shuffle the dataset
shuffle(images)

# Split into train and validation
train_count = int(len(images) * train_ratio)
train_files = images[:train_count]
val_files   = images[train_count:]

print(f"Total images: {len(images)}")
print(f"Train images: {len(train_files)}")
print(f"Validation images: {len(val_files)}")

# Function to copy files
def copy_files(file_list, img_dst, lbl_dst):
    for img_file in file_list:
        label_file = img_file.rsplit(".", 1)[0] + ".txt"  # corresponding label file
        img_src = os.path.join(source_path, img_file)
        lbl_src = os.path.join(source_path, label_file)

        # Copy image
        shutil.copy(img_src, os.path.join(img_dst, img_file))

        # Copy label if it exists
        if os.path.exists(lbl_src):
            shutil.copy(lbl_src, os.path.join(lbl_dst, label_file))
        else:
            print(f"⚠️ Label file missing for {img_file}")

# Copy train and validation files
copy_files(train_files, train_image_path, train_label_path)
copy_files(val_files, val_image_path, val_label_path)

print("✅ Dataset split completed!")

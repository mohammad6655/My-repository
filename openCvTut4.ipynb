```python
import numpy as np
import cv2 as cv
```

```python
img = cv.imread('E:\\New folder (6)\\CC7C8857-F56B-4524-8E66-A5DD0D03B16F-540x800 (1).jpeg')
```

```python
imgray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)
```

```python
ret, thresh = cv.threshold(imgray, 44, 255, cv.THRESH_BINARY)
```

```python
ret2, thresh2 = cv.threshold(imgray, 0, 255, cv.THRESH_BINARY+cv.THRESH_OTSU)
```

```python
thresh3 = cv.adaptiveThreshold(imgray, 255, cv.ADAPTIVE_THRESH_GAUSSIAN_C, cv.THRESH_BINARY, 155, 19)
```

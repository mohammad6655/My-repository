```python
from imageai.Detection import ObjectDetection, VideoObjectDetection
import os
```

# Image object detection

```python
model_path = "E:\\YOLOv3\\"
image_path = "E:\\image\\"
```

```python
detector = ObjectDetection()
detector.setModelTypeAsYOLOv3()
```

```python
detector.setModelPath(os.path.join(model_path, "yolov3.pt"))
```

```python
detector.loadModel()
```

```python
detections = detector.detectObjectsFromImage(input_image=os.path.join(image_path,"im.jpg"), output_image_path = os.path.join(image_path,"im_detected.jpg"))
```

```python
custom_detector = detector.CustomObjects(person=True)
```

```python
custom_detection = detector.detectObjectsFromImage(custom_objects=custom_detector,input_image=os.path.join(image_path,"im.jpg"), output_image_path = os.path.join(image_path,"im_detected_dog.jpg"))
```

```python
custom_detection
```

```python
import cv2 as cv

image = cv.imread("E:\\image\\im_detected_dog.jpg")

for detection in custom_detection:
    if detection["name"] == "person":
        box_points = detection["box_points"]
        cv.rectangle(image, (box_points[0], box_points[1]), (box_points[2], box_points[3]), (255, 0, 0), 2)
        cv.putText(image, f"{detection['name']} {int(detection['percentage_probability'])}%", 
                    (box_points[0], box_points[1] - 10), cv.FONT_HERSHEY_SIMPLEX, 0.5, (255, 0, 0), 2)

cv.imwrite("E:\image\\im_custom_colored.jpg", image)
cv.imshow("E:\image\\im_custom_colored.jpg", image)
cv.waitKey(0)
cv.destroyAllWindows()
```
